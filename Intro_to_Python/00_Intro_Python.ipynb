### 📦 What does `import pandas as pd` mean?

When we write:

```python
import pandas as pd
```

we are doing three things:

1. **`import`**  
   This tells Python to load an external library so we can use its tools and functions.

2. **`pandas`**  
   This is a library that helps us work with structured data like tables, spreadsheets, or CSV files. It provides the **DataFrame** object, which allows us to store, explore, and manipulate data efficiently.

3. **`as pd`**  
   This creates a shortcut or alias. Instead of writing `pandas` every time, we can just write `pd`.  
   Example: `pd.read_csv("file.csv")` instead of `pandas.read_csv("file.csv")`.

---

### ✅ Summary

```python
import pandas as pd
```

> Load the `pandas` library and give it the short name `pd` so we can use it easily in our code.

This is a standard and recommended practice in Python data analysis.

### 📊 Plotting with `matplotlib.pyplot`

To create charts in Python, we use a library called **matplotlib**. One part of this library, called `pyplot`, is especially useful for making simple plots like bar charts, line charts, and scatter plots.

---

### 🔍 Line-by-line explanation

```python
import matplotlib.pyplot as plt
```
This line **loads** the `pyplot` module from the `matplotlib` library and gives it the shortcut name `plt`.  
We'll use `plt` to access the plotting functions.

---

```python
plt.bar(df["Company"], df["Price"])
```
This line **creates a bar chart**:

- `df["Company"]` gives the labels for the x-axis (e.g., company names).
- `df["Price"]` gives the values for the y-axis (e.g., stock prices).

---

```python
plt.title("Stock Prices")
```
Adds a **title** to the chart: `"Stock Prices"`.

---

```python
plt.show()
```
This line **displays** the chart. Without it, the chart might not appear in some environments.

---

### ✅ Summary

These lines together create and show a bar chart that compares stock prices across different companies using data stored in a DataFrame called `df`.



````python
import matplotlib.pyplot as plt
plt.bar(df["Company"], df["Price"])
plt.title("Stock Prices")
plt.show()
````

````python
# 0.6 – Simple Data Transformations
## Adding a new column
exchange_rate = 1.08  # EUR to USD
df["Price_USD"] = df["Price"] * exchange_rate
print(df)
````
