```python
## DataFrame using pandas
import pandas as pd
data = {"Company": ["Santander", "BBVA", "Mapfre"], "Price": [3.77, 6.21, 1.95]}
df = pd.DataFrame(data)
print(df)
```

```python
# ✅ Exercise 0.3:
# Add a new company and price to the DataFrame. Example: "CaixaBank", 4.12
data = {"Company": ["Santander", "BBVA", "Mapfre","CaixaBank"], "Price": [3.77, 6.21, 1.95, 4.12]}
df = pd.DataFrame(data)
print(df)
```

```python
# 0.4 – Reading and Writing Files
## Upload a CSV manually or use a public URL
df.to_csv("example.csv", index=False)
df2 = pd.read_csv("example.csv")
print(df2.head())
```

```python
# 0.5 – Basic Statistics and Visualizations
print(df["Price"].mean())
print(df["Price"].max())
print(df["Price"].min())
print(df["Price"].std())
```

```python
import matplotlib.pyplot as plt
plt.bar(df["Company"], df["Price"])
plt.title("Stock Prices")
plt.show()
```

```python
# 0.6 – Simple Data Transformations
## Adding a new column
exchange_rate = 1.08  # EUR to USD
df["Price_USD"] = df["Price"] * exchange_rate
print(df)
```

```python
# ✅ Exercise 0.6:
# Try using an exchange rate of 1.10 instead. How does it change the results?
exchange_rate = 1.10  # EUR to USD
df["Price_USD"] = df["Price"] * exchange_rate
print(df)
```
